# Task 15: Clinic Appointment Manager

```python
patients = []

def add_patient():
    patient_id = input("Enter Patient ID: ")
    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient ID already exists!")
            return

    name = input("Enter Patient Name: ")
    age = int(input("Enter Patient Age: "))
    patient_type = input("Enter Patient Type (normal/emergency): ")

    patient = {
        "id": patient_id,
        "name": name,
        "age": age,
        "type": patient_type.lower()
    }

    patients.append(patient)

    print("Patient added successfully!")

def search_patient():
    patient_id = input("Enter Patient ID: ")

    found = False
    for patient in patients:
        if patient["id"] == patient_id:
            print("\nPatient Found")
            print("ID:", patient["id"])
            print("Name:", patient["name"])
            print("Age:", patient["age"])
            print("Type:", patient["type"])
            found = True
            break

    if not found:
        print("Patient not found!")

def display_patients():
    if not patients:
        print("No patients available.")
        return

    print("\n===== PATIENT LIST =====")

    # Display Emergency Patients
    emergency_found = False
    print("\nEmergency Patients:")
    for patient in patients:
        if patient["type"] == "emergency":
            print(
                "ID:", patient["id"],
                "| Name:", patient["name"],
                "| Age:", patient["age"]
            )
            emergency_found = True
    if not emergency_found:
        print("No emergency patients.")

    # Display Normal Patients
    normal_found = False
    print("\nNormal Patients:")
    for patient in patients:
        if patient["type"] == "normal":
            print(
                "ID:", patient["id"],
                "| Name:", patient["name"],
                "| Age:", patient["age"]
            )
            normal_found = True
    if not normal_found:
        print("No normal patients.")

def cancel_appointment():
    patient_id = input("Enter Patient ID: ")

    original_len = len(patients)
    # Use list comprehension to create a new list without the patient to cancel
    global patients
    patients = [p for p in patients if p["id"] != patient_id]

    if len(patients) < original_len:
        print("Appointment cancelled successfully!")
    else:
        print("Patient ID not found!")

def count_patients():
    normal = 0
    emergency = 0

    for patient in patients:
        if patient["type"] == "normal":
            normal += 1
        elif patient["type"] == "emergency":
            emergency += 1

    print("\n===== PATIENT COUNT =====")
    print("Normal Patients:", normal)
    print("Emergency Patients:", emergency)
    print("Total Patients:", normal + emergency)

while True:
    print("\n===== PATIENT MANAGEMENT SYSTEM =====")
    print("1. Add Patient")
    print("2. Search Patient")
    print("3. Display All Patients")
    print("4. Cancel Appointment")
    print("5. Show Patient Count")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_patient()
    elif choice == "2":
        search_patient()
    elif choice == "3":
        display_patients()
    elif choice == "4":
        cancel_appointment()
    elif choice == "5":
        count_patients()
    elif choice == "6":
        print("Thank you!")
        break
    else:
        print("Invalid choice!")
```
